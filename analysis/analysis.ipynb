```python
import os
import json
import glob

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Better plot style
sns.set_style("whitegrid")

print("Libraries imported successfully.")
```

```python
import glob
import json
import pandas as pd

records = []

for file in glob.glob("data/*.json"):
    with open(file, "r", encoding="utf-8") as f:
        data = json.load(f)

        # JSON array ဖြစ်ရင်
        if isinstance(data, list):
            records.extend(data)

        # JSON object တစ်ခုချင်းဖြစ်ရင်
        else:
            records.append(data)

df = pd.DataFrame(records)

print(f"Total Records : {len(df)}")
print(f"Total Columns : {len(df.columns)}")

df.head()
```

```python
print("Dataset Shape:", df.shape)
print()

df.info()

print("Missing Values")

df.isnull().sum()

print("Before:", len(df))

df = df.dropna(subset=["latency_ms"])

print("After :", len(df))

print("Duplicate Records:", df.duplicated().sum())

df = df.drop_duplicates()

print("Remaining:", len(df))

df["timestamp"] = pd.to_datetime(df["timestamp"])

df.dtypes

df["date"] = df["timestamp"].dt.date
df["hour"] = df["timestamp"].dt.hour
df["weekday"] = df["timestamp"].dt.day_name()

df.head()
```

```python
# Import plotting library (run once)
import matplotlib.pyplot as plt
```

```python
# Statistical summary
df.describe()
# Provider distribution
df["provider"].value_counts()
# Connection type distribution
df["connection_type"].value_counts()
# Operating system distribution
df["os_platform"].value_counts()
# Routing region distribution
df["routing_region"].value_counts()
# Location distribution
df["location_zone"].value_counts()
# Location distribution
df["location_zone"].value_counts()
# Connection status distribution
df["status"].value_counts()
```

```python
connected_df = df[df["status"] == "Connected"]

provider_counts = connected_df["provider"].value_counts()

plt.figure(figsize=(8,5))

provider_counts.plot(kind="bar")

plt.title("Internet Service Provider Distribution (Connected Only)")
plt.xlabel("Provider")
plt.ylabel("Number of Records")

plt.xticks(rotation=0)

plt.show()
```

```python
connected_df = df[df["status"] == "Connected"].copy()
connected_df["provider"].value_counts()

avg_latency_provider = (
    connected_df.groupby("provider")["latency_ms"]
    .mean()
    .sort_values()
)

avg_latency_provider
```

```python
plt.figure(figsize=(8,5))

avg_latency_provider.plot(kind="bar")

plt.title("Average Latency by ISP (Connected Only)")
plt.xlabel("ISP Provider")
plt.ylabel("Average Latency (ms)")

plt.xticks(rotation=0)

plt.show()
```
